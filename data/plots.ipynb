```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']
```

```python
df = pd.read_csv('proc-data.csv')
```

```python
df
```

```python
participants = df['Participant ID'].unique()
```

```python
plt.figure(figsize=(10, 6))

for i, p in enumerate(participants):
    # Filter and sort by the x-axis (Visual Latency)
    subset = df[df['Participant ID'] == p].sort_values('Visual Latency')
    plt.plot(subset['Visual Latency'], subset['Normalized WPM'], 
             marker='o', label=p, linewidth=2, color=colors[i])

plt.title('Normalized WPM vs Visual Latency', fontsize=14)
plt.xlabel('Visual Latency (ms)', fontsize=12)
plt.ylabel('Normalized WPM', fontsize=12)
plt.xticks(df['Visual Latency'].unique()) # Show all latency steps on X-axis
plt.legend(title='Participant')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

for i, p in enumerate(participants):
    # Filter and sort by the x-axis (Sample Index)
    subset = df[df['Participant ID'] == p].sort_values('Sample Index')
    plt.plot(subset['Sample Index'], subset['Normalized WPM'], 
             marker='o', label=p, linewidth=2, color=colors[i])

plt.title('Normalized WPM vs Sample Index', fontsize=14)
plt.xlabel('Sample Index (Order of Appearance)', fontsize=12)
plt.ylabel('Normalized WPM', fontsize=12)
plt.xticks(df['Sample Index'].unique()) # Show all sample indices as whole numbers
plt.legend(title='Participant')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# 1. Average across participants for each Visual Latency
avg_latency_df = df.groupby('Visual Latency')['Normalized WPM'].mean().reset_index()
avg_latency_df = avg_latency_df.sort_values('Visual Latency')

# 2. Average across participants for each Sample Index
avg_sample_df = df.groupby('Sample Index')['Normalized WPM'].mean().reset_index()
avg_sample_df = avg_sample_df.sort_values('Sample Index')

# Optional: Display the aggregated data to verify
print("Average by Latency:")
display(avg_latency_df)
print("\nAverage by Sample Index:")
display(avg_sample_df)
```

```python
plt.figure(figsize=(10, 6))

# Plot the average line by Latency
plt.plot(avg_latency_df['Visual Latency'], avg_latency_df['Normalized WPM'], 
         marker='o', color='black', linewidth=3, markersize=8, label='Average')

plt.title('Average Normalized WPM vs Visual Latency', fontsize=14)
plt.xlabel('Visual Latency (ms)', fontsize=12)
plt.ylabel('Average Normalized WPM', fontsize=12)
plt.xticks(df['Visual Latency'].unique()) 
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# Plot the average line by Sample Index
plt.plot(avg_sample_df['Sample Index'], avg_sample_df['Normalized WPM'], 
         marker='o', color='black', linewidth=3, markersize=8, label='Average')

plt.title('Average Normalized WPM vs Sample Index', fontsize=14)
plt.xlabel('Sample Index (Order of Appearance)', fontsize=12)
plt.ylabel('Average Normalized WPM', fontsize=12)
plt.xticks(df['Sample Index'].unique())
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Group by Visual Latency and calculate the mean Comfort across all participants
avg_comfort_df = df.groupby('Visual Latency')['Comfort'].mean().reset_index()

# Sort by Latency to ensure the line plots correctly
avg_comfort_df = avg_comfort_df.sort_values('Visual Latency')

# Optional: Display the aggregated data
display(avg_comfort_df)
```

```python
plt.figure(figsize=(10, 6))

# Plot the average comfort line (black, thicker line, triangle marker to match previous comfort plots)
plt.plot(avg_comfort_df['Visual Latency'], avg_comfort_df['Comfort'], 
         marker='^', color='black', linewidth=3, markersize=8, label='Average Comfort')

plt.title('Average Comfort vs Visual Latency', fontsize=14)
plt.xlabel('Visual Latency (ms)', fontsize=12)
plt.ylabel('Average Comfort Rating (1-4)', fontsize=12)
plt.xticks(df['Visual Latency'].unique()) 
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Group by Sample Index and calculate the mean Comfort across all participants
avg_comfort_sample_df = df.groupby('Sample Index')['Comfort'].mean().reset_index()

# Sort by Sample Index to ensure the line plots correctly from left to right
avg_comfort_sample_df = avg_comfort_sample_df.sort_values('Sample Index')

# Optional: Display the aggregated data
display(avg_comfort_sample_df)
```

```python
plt.figure(figsize=(10, 6))

# Plot the average comfort line (black, thicker line, triangle marker to match previous comfort plots)
plt.plot(avg_comfort_sample_df['Sample Index'], avg_comfort_sample_df['Comfort'], 
         marker='^', color='black', linewidth=3, markersize=8, label='Average Comfort')

plt.title('Average Comfort vs Sample Index', fontsize=14)
plt.xlabel('Sample Index (Order of Appearance)', fontsize=12)
plt.ylabel('Average Comfort Rating (1-4)', fontsize=12)
plt.xticks(df['Sample Index'].unique()) 
plt.yticks([1, 2, 3, 4]) # Ensures only whole numbers appear on the Y-axis
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
